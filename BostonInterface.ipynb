```python
import gradio as gr
import numpy as np
from tensorflow.keras.models import load_model
from gtts import gTTS
import os

# Load the trained model
model = load_model("boston_housing_model.h5")

# Function to make predictions
def predict_price(*features):
    features = np.array(features).reshape(1, -1)  # Convert input to NumPy array
    prediction = model.predict(features)[0][0]  # Get the predicted value
    text = f"The predicted price is {prediction:.2f}"
    
    # Convert text to speech
    tts = gTTS(text=text, lang='en')
    tts.save("prediction.mp3")  # Save as MP3 file
    
    return text, "prediction.mp3"

# Define UI with Gradio
inputs = [gr.Number(label=f"Feature {i+1}") for i in range(7)]  # Assuming 13 features
interface = gr.Interface(fn=predict_price, inputs=inputs, outputs=["text", "audio"])

# Launch the app
interface.launch()
```
